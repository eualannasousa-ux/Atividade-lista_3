```python
numeros = []
for i in range(10):
    n = int(input(f"Digite o {i+1}º número: "))
    numeros.append(n)

pares = []
impares = []
for n in numeros:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)

print("Todos:", numeros)
print("Pares:", pares)
print("Ímpares:", impares)
```

```python
numeros = []
for i in range(8):
    n = int(input(f"Digite o {i+1}º número: "))
    numeros.append(n)

maior = numeros[0]
menor = numeros[0]
pos_maior = 0
pos_menor = 0

for i in range(len(numeros)):
    if numeros[i] > maior:
        maior = numeros[i]
        pos_maior = i
    if numeros[i] < menor:
        menor = numeros[i]
        pos_menor = i

print("Lista:", numeros)
print("Maior:", maior, "posição:", pos_maior)
print("Menor:", menor, "posição:", pos_menor)
```

```python
nomes = []
while True:
    nome = input("Digite um nome (FIM para sair): ")
    if nome == "FIM":
        break
    if nome not in nomes:
        nomes.append(nome)

print("Nomes:", nomes)
print("Quantidade:", len(nomes))
```

```python
valores = []
for i in range(6):
    n = int(input(f"Digite o {i+1}º número: "))
    valores.append(n)
tupla = tuple(valores)

cont5 = tupla.count(5)
pares = 0
for n in tupla:
    if n % 2 == 0:
        pares += 1

print("Tupla:", tupla)
print("Vezes que o 5 apareceu:", cont5)
if cont5 > 0:
    print("Primeira posição do 5:", tupla.index(5))
else:
    print("O número 5 não apareceu")
print("Quantidade de pares:", pares)
```

```python
aluno = {}
aluno["nome"] = input("Nome: ")
aluno["turma"] = input("Turma: ")
aluno["nota1"] = float(input("Nota1: "))
aluno["nota2"] = float(input("Nota2: "))

media = (aluno["nota1"] + aluno["nota2"]) / 2
aluno["media"] = media

if media >= 7:
    aluno["situacao"] = "aprovado"
elif media >= 5:
    aluno["situacao"] = "recuperação"
else:
    aluno["situacao"] = "reprovado"

print(aluno)
```

```python
frequencia = {}
for i in range(8):
    palavra = input(f"Digite a {i+1}ª palavra: ")
    if palavra in frequencia:
        frequencia[palavra] += 1
    else:
        frequencia[palavra] = 1

for palavra, qtd in frequencia.items():
    print(palavra, ":", qtd)
```

```python
estoque = {"arroz": 10, "feijão": 20, "açúcar": 15, "café": 5}

produto = input("Produto: ")
qtd_vendida = int(input("Quantidade vendida: "))

if produto not in estoque:
    print("Produto não existe")
elif estoque[produto] < qtd_vendida:
    print("Estoque insuficiente")
else:
    estoque[produto] -= qtd_vendida

print("Estoque final:", estoque)
```

```python
turma = []
for i in range(5):
    nome = input("Nome: ")
    n1 = float(input("Nota1: "))
    n2 = float(input("Nota2: "))
    media = (n1 + n2) / 2
    aluno = {"nome": nome, "notas": (n1, n2), "media": media}
    turma.append(aluno)

for aluno in turma:
    print(aluno["nome"], "-", aluno["media"])
```

```python
estoque = []
for i in range(5):
    codigo = int(input("Código: "))
    nome = input("Nome: ")
    qtd = int(input("Quantidade: "))
    preco = float(input("Preço: "))
    item = {"codigo": codigo, "nome": nome, "quantidade": qtd, "preco": preco}
    estoque.append(item)

# (a)
print("Itens:", estoque)

# (b) e (c)
total_estoque = 0
maior_item = estoque[0]
maior_valor = estoque[0]["quantidade"] * estoque[0]["preco"]

for item in estoque:
    valor = item["quantidade"] * item["preco"]
    total_estoque += valor
    if valor > maior_valor:
        maior_valor = valor
        maior_item = item

print("Valor total do estoque:", total_estoque)
print("Item de maior valor:", maior_item)

# (d)
resumo = (len(estoque), total_estoque)
print("Resumo (qtd_itens, valor_total):", resumo)
```
